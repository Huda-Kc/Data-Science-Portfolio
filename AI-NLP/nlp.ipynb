```python
import re
import string

def wordopt(text):
    # Convert to string just in case
    text = str(text).lower()
    
    # 1. Remove URLs / Website links
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    
    # 2. Remove HTML tags
    text = re.sub(r'<.*?>', '', text)
    
    # 3. Remove text inside square brackets [like this]
    text = re.sub(r'\[.*?\]', '', text)
    
    # 4. Remove punctuation marks
    text = re.sub('[%s]' % re.escape(string.punctuation), '', text)
    
    # 5. Remove newlines
    text = re.sub('\n', ' ', text)
    
    # 6. Remove words containing numbers (e.g., th1s, 3d)
    text = re.sub(r'\w*\d\w*', '', text)
    
    # 7. CRITICAL FIX: Keep letters, replace ONLY special symbols/non-words with spaces
    # (Using capital \W instead of lowercase \w)
    text = re.sub(r'\W', ' ', text)
    
    return text
```

```python
df['text']=df['text'].apply(wordopt)
```

```python
# BOW
list1=["Dogs bites man","Man bites dog.","Dog eats meat","Man eats food"]
list1

processed_list=[doc.lower().replace(".","")for doc in list1]
processed_list

print("Our corpus:",processed_list)
```

```python
# countvectorizer
from sklearn.feature_extraction.text import CountVectorizer
vect=CountVectorizer()

# Build a BOW representsion for the corpus
count_bow=vect.fit_transform(processed_list)
print("our vocabulary:",vect.vocabulary_)
```

```python
print("BoW representation for 'dog bites man': ", count_bow[0].toarray()) #ist sentence
print("BoW representation for 'man bites dog':  ",count_bow[1].toarray()) #2nd sentence
print("BoW representation for 'dog eats meat':  ",count_bow[2].toarray()) 
print("BoW representation for 'man eats food':  ",count_bow[3].toarray()) 
```

```python
# binary vector
count_vec =CountVectorizer(binary=True)
count_vec.fit(processed_list)
temp=count_vec.transform(["cat and dog are thick friends"])
print("Bow representation for 'cat and dog are thick friends':",temp.toarray())
```
