```python
import tensorflow as tf
from sklearn.model_selection import train_test_split
from tensorflow.keras.layers import Input,Conv2D,MaxPool2D,Dense,Dropout,Flatten,Activation
from tensorflow.keras.models import Sequential
from tensorflow.keras.losses import sparse_categorical_crossentropy
from tensorflow.keras.metrics import sparse_top_k_categorical_accuracy
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.datasets import fashion_mnist
```

```python
(x_train,y_train),(x_test,y_test)=fashion_mnist.load_data()
```

```python
x_train
```

```python
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

```python
x_train=x_train.reshape(60000,28,28,1)
x_test=x_test.reshape(10000,28,28,1)
```

```python
x_train=x_train /255.0
x_test=x_test / 255.0
```

```python
x_train,x_val,y_train,y_val=train_test_split(x_train,y_train,test_size=0.2,random_state=42)
```

```python
print(x_train.shape)
print(y_train.shape)
print(x_val.shape)
print(y_val.shape)
print(x_test.shape)
print(y_test.shape)
```

```python
cnn=Sequential([
    Input(shape=(28,28,1)),

    Conv2D(256,(3,3)),
    MaxPool2D(2,2),
    Dropout(0.2),

    Conv2D(128,(3,3)),
    MaxPool2D(2,2),
    Flatten(),

    Dense(128,activation='relu'),
    Dense(64,activation='relu'),
    Dense(32,activation='relu'),
    Dense(10,activation='relu')
])
```

```python
model=cnn.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
early_stop=EarlyStopping(monitor='val_accuracy',mode='max',verbose=1,patience=10,restore_best_weights=True)
```

```python
cnn.fit(x_train,y_train,
        epochs=100,
        batch_size=500,
        validation_data=(x_test,y_test),
        callbacks=[early_stop])
```

```python
loss,accuracy=model.evaluate(x_test,y_test)
print("test accuracy:",accuracy)
```
